```python
def extraire_texte_pdf(chemin_pdf: str) -> str:
    """
    Fonction simple pour lire toutes les pages d'un fichier PDF
    et en extraire le texte brut.
    """
    texte_complet = []
    try:
        with open(chemin_pdf, "rb") as f:
            lecteur = pypdf.PdfReader(f)
            for page_num in range(len(lecteur.pages)):
                page = lecteur.pages[page_num]
                texte_page = page.extract_text()
                if texte_page:
                    texte_complet.append(texte_page)
        return "\n".join(texte_complet)
    except Exception as e:
        print(f"Erreur lors de la lecture du PDF : {e}")
        return ""

# --- Exemple d'utilisation (Commenté tant que vous n'avez pas téléversé de fichier) ---
# 1. Téléversez votre fichier PDF dans Colab (via l'icône de dossier à gauche)
# 2. Décommentez les lignes ci-dessous et remplacez 'votre_fichier.pdf' par le nom de votre fichier

# chemin_du_document = "votre_fichier.pdf"
# texte_extrait = extraire_texte_pdf(chemin_du_document)
# if texte_extrait:
#     # Découper le texte extrait de votre PDF en segments
#     segments = chunker.split_text(texte_extrait)
#     print(f"Votre PDF a été découpé en {len(segments)} segments.")
```

### Étape 1 : Ingestion et Découpage (Loader & Chunker)

Définir une classe simple pour simuler ou charger des documents et les découper en  (chunks) avec un  (overlap) ajustable.

```python
class SimpleChunker:
    def __init__(self, chunk_size: int = 150, chunk_overlap: int = 20):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap

    def split_text(self, text: str) -> list:

        text = re.sub(r'\s+', ' ', text).strip()
        words = text.split(' ')
        chunks = []

        i = 0
        while i < len(words):

            chunk_words = words[i:i + self.chunk_size]
            chunk_text = " ".join(chunk_words)
            if chunk_text:
                chunks.append(chunk_text)

            i += (self.chunk_size - self.chunk_overlap)
            if self.chunk_size <= self.chunk_overlap:
                break

        return chunks

chunker = SimpleChunker(chunk_size=15, chunk_overlap=3)
exemple_texte = "Le Retrieval-Augmented Generation (RAG) est une technique qui combine la recherche d'informations et la génération de texte par un modèle de langage."
segments = chunker.split_text(exemple_texte)

for idx, seg in enumerate(segments):
    print(f"Segment {idx+1} : {seg}")
```

### Étape 2 : Embeddings locaux (Sentence Transformers)

Pour transformer ces segments de texte en représentations vectorielles, nous utilisons directement la bibliothèque `sentence-transformers` sans surcouche.

```python
class LocalEmbeddings:
    def __init__(self, model_name: str = "all-MiniLM-L6-v2"):
        self.model = SentenceTransformer(model_name)

    def get_embeddings(self, texts: list) -> list:
        embeddings = self.model.encode(texts, show_progress_bar=False)
        return [emb.tolist() for emb in embeddings]

embedder = LocalEmbeddings()
vecteurs = embedder.get_embeddings(segments)
print(f"Génération réussie de {len(vecteurs)} vecteurs. Dimension de chaque vecteur : {len(vecteurs[0])}")
```

### Étape 3 : Base de données vectorielle locale (ChromaDB)
Nous utilisons le client natif de `chromadb` pour créer une collection persistante, y stocker nos segments de texte ainsi que leurs représentations vectorielles personnalisées (embeddings).

```python
chroma_client = chromadb.PersistentClient(path="./chroma_db")

collection_name = "mon_rag_local"
collection = chroma_client.get_or_create_collection(name=collection_name)

ids = [f"doc_{i}" for i in range(len(segments))]

collection.add(
    embeddings=vecteurs,
    documents=segments,
    ids=ids
)

print(f"Stockage réussi ! {collection.count()} segments ont été indexés dans la base de données vectorielle.")
```

### Étape 4 : Recherche et Récupération par similarité (Retrieval)
Pour répondre à une question, nous devons d'abord la convertir en vecteur d'embedding avec notre modèle local, puis rechercher dans notre base ChromaDB les segments les plus proches géométriquement.

```python
requete_utilisateur = "    "

vecteur_requete = embedder.get_embeddings([requete_utilisateur])[0]

resultats_cv = collection.query(
    query_embeddings=[vecteur_requete],
    n_results=40
)

print(f"Question posée : '{requete_utilisateur}'\n")
print("--- 4 Segments pertinents trouvés avec la bonne requête ---")
for i, doc in enumerate(resultats_cv['documents'][0]):
    dist = resultats_cv['distances'][0][i]
    #print(f"\nSegment {i+1} (Distance : {dist:.4f}) :\n=> \"{doc}\"")
```
